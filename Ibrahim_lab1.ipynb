```python
print("=== AI TRAVEL AGENT ===")

budget = float(input("Enter your budget ($): "))
season = input("Enter travel season (summer/winter/rainy): ").strip().lower()
trip_type = input("What kind of trip? (beach/mountain/city): ").strip().lower()
days = int(input("How many days are you traveling? "))

if trip_type == "beach":
    if season == "summer":
        destination = "Maldives"
    elif season == "winter":
        destination = "Zanzibar"
    else:
        destination = "Goa, India"

elif trip_type == "mountain":
    if season == "winter":
        destination = "Swiss Alps"
    elif season == "summer":
        destination = "Rocky Mountains"
    else:
        destination = "Scottish Highlands"

elif trip_type == "city":
    if season == "winter":
        destination = "New York"
    elif season == "summer":
        destination = "Paris"
    else:
        destination = "Tokyo"

else:
    destination = "Not sure, please choose beach, mountain, or city"

if budget < 500:
    budget_note = "Budget is low, consider a nearby domestic destination instead."
elif budget < 1500:
    budget_note = "Budget is moderate, a standard package is recommended."
else:
    budget_note = "Budget is high, you can afford a premium package."

if days <= 3:
    length_note = "Short trip - a quick weekend getaway plan will be created."
elif days <= 7:
    length_note = "Medium trip - a full week itinerary will be created."
else:
    length_note = "Long trip - an extended vacation plan will be created."

print("\n--- TRAVEL RECOMMENDATION ---")
print("Suggested Destination:", destination)
print("Budget Note:", budget_note)
print("Trip Length Note:", length_note)
```
