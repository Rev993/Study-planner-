```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="AQ.Ab8RN6JNcJk_n4jj2ck42gstcSyk0TLOFWVH7FbZQFlw2FAEOw")

MODEL = "models/gemini-3.5-flash"

semester_6_section_b = {
    "AI": "Miss Uzma",
    "NSC": "Akhtar Raza",
    "CC": "Miss nahy",
    "PP": "Dr. Sarim",
    "NLP": "X",
}


def get_subject_info(subject_code):
    subject_code = subject_code.upper().strip()
    if subject_code in semester_6_section_b:
        teacher = semester_6_section_b[subject_code]
        return f"{subject_code} is taught by {teacher} in Semester 6, Section B."
    else:
        return None


def list_all_subjects():
    result = "Semester 6 - Section B Subjects:\n"
    for subject, teacher in semester_6_section_b.items():
        result += f"- {subject}: {teacher}\n"
    return result


def info_agent(user_input):
    text = user_input.lower()

    if "all subjects" in text or "list subjects" in text or "which subjects" in text:
        return list_all_subjects()

    for subject_code in semester_6_section_b:
        if subject_code.lower() in text:
            return get_subject_info(subject_code)

    prompt = f"""
You are an information assistant for Federal Urdu University (FUUAST).

Answer the student's query clearly and concisely.
If it's about courses, teachers, or academic info you don't know, say you don't have that data.

User request: {user_input}

Answer:
"""
    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )
    return response.text


print("FUUAST Info Assistant Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = info_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
