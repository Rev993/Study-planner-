```python
def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Pakistan tourism.

Provide a well-structured travel plan with clear headings:

1. Destination Overview
2. Budget Travel Options (Bus, Train, Flight with approx cost in PKR)
3. Places to Visit (specific locations)
4. Food Suggestions (local dishes)
5. Money-Saving Tips

Make response detailed but easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
def study_planner_agent(user_input):

    prompt = f"""
You are an AI Study Planner Agent that helps students organize their study schedule.

Analyze the subjects, deadlines, difficulty levels, and available study time provided by the student.

Provide a well-structured study plan with clear headings:

1. Subject Overview (list subjects with their deadlines and difficulty)
2. Priority Ranking (which subjects need attention first, and why)
3. Daily/Weekly Study Schedule (specific time blocks per subject)
4. Study Tips (based on difficulty level of each subject)
5. Adjustment Notes (what to do if a deadline changes or time runs short)

Make response detailed but easy for students to follow.

Student request: {user_input}

Study Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("AI Study Planner Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = study_planner_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
